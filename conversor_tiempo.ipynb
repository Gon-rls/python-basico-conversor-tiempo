Ejercicio: Conversor de tiempo con visualización

Se desea implementar una función llamada conversor_de_tiempo que permita convertir una cantidad de tiempo dada en segundos a una unidad especificada por el usuario: minutos, horas o días. Además, la función debe mostrar una visualización gráfica tipo gráfico de barras que compare el valor original (en segundos) con el valor convertido.

La función debe cumplir con los siguientes requisitos:

1 Validar que el valor de entrada en segundos sea numérico (entero o flotante) y no negativo.

2 Validar que la unidad de destino sea una de las siguientes: 'minutos', 'horas', 'días'.

3 Realizar la conversión correspondiente según la unidad de destino.

4 Generar un gráfico de barras que represente visualmente la cantidad en
segundos y su conversión.

5 Retornar un mensaje de texto indicando el resultado de la conversión con dos decimales de precisión.

```
# Importamos la librería para crear gráficos que se nos piden en el ej "conversor_de_tiempo"
import matplotlib.pyplot as plt

# Definimos la función tal cual se nos pide en el ejercicio y la llamamos como se nos pide
def conversor_de_tiempo(segundos, unidad_destino):

# Paso 1: Validamos que 'segundos' sea un número válido y no negativo ya que si es negativo no funcionara ademas de que sea entero o flotante.
    try:
        segundos = float(segundos)  # Convertimos a número decimal
        if segundos < 0:  # Si es negativo, mostramos error
            return "El número de segundos no puede ser negativo"
    except:  # Si no es un número, mostramos otro mensaje
        return "Debes ingresar un número válido"

# Paso 2: Verificamos que la unidad sea válida
    if unidad_destino not in ["minutos", "horas", "días"]:
        return "La unidad debe ser 'minutos', 'horas' o 'días'"

# Paso 3: Realizamos la conversión según la unidad elegida 1 minuto son 60 segundos, 1 hora son 3600 segundos, y 60 minutos son 1 hora. 1 dia son 86400 segundos o 24 horas.
    if unidad_destino == "minutos":
        convertido = segundos / 60
    elif unidad_destino == "horas":
        convertido = segundos / 3600
    elif unidad_destino == "días":
        convertido = segundos / 86400

  # Paso 4: Creamos el gráfico de barras con los dos valores que se nos pide

    valores = [segundos, convertido]  # Lista con los valores a comparar
    etiquetas = ["Segundos", unidad_destino.capitalize()]  # Nombres de las barras en el grafico
    plt.bar(etiquetas, valores, color=["blue", "green"])  # Creamos el gráfico
    plt.title("Conversión de tiempo")  # Título del gráfico saldra arriba del todo
    plt.ylabel("Cantidad")  # Etiqueta del eje Y
    plt.show()  # Mostramos el gráfico para que se vea adecuadamente.

# Paso 5: Mostramos el resultado final con dos decimales
    return f"{segundos:.2f} segundos equivalen a {convertido:.2f} {unidad_destino}"

# Prueba del programa con 3600 segundos a horas
print(conversor_de_tiempo(3600, "horas"))
```
